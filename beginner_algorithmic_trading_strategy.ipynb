# Beginner Algorithmic Trading Strategy
This notebook simulates a simple strategy using sample market prices. Run each code cell from top to bottom.

**How to run a cell:** Select it and press `Shift + Enter`.

## Block 1 — Input Data (The Market)
We start with a list of eight sample closing prices. `len(prices)` tells us how many prices are in the list.

```python
prices = [100, 102, 105, 110, 108, 107, 115, 120]
print(f"Market Data Loaded: {len(prices)} days.")
```

## Block 2 — Indicator (The Average)
`sum(prices)` adds all prices together. `len(prices)` counts them. Dividing the total by the count gives the arithmetic mean.

```python
total = sum(prices)
count = len(prices)
average = total / count

print(f"The Average Price is: ${average:.2f}")
```

## Block 3 — Decision Engine (The Loop)
The loop checks every price. If the price is above the average, it prints `BUY`; otherwise, it prints `WAIT`.

```python
print("--- Starting Strategy Scan ---")

for p in prices:
    if p > average:
        print(f"Price ${p} is > Average. Signal: BUY")
    else:
        print(f"Price ${p} is <= Average. Signal: WAIT")

print("--- Scan Complete ---")
```

## Block 4 — Final Check
This cell checks the calculated average and confirms which prices are above it.

```python
print(f"Calculated average: ${average:.3f}")

above_average = [p for p in prices if p > average]
at_or_below_average = [p for p in prices if p <= average]

print("Prices above average (BUY):", above_average)
print("Prices at or below average (WAIT):", at_or_below_average)

assert len(prices) == 8
assert round(average, 3) == 108.375
assert above_average == [110, 115, 120]
assert at_or_below_average == [100, 102, 105, 108, 107]

print("All checks passed!")
print(f"Calculated average: ${average:.3f}")

above_average = [p for p in prices if p > average]
at_or_below_average = [p for p in prices if p <= average]

print("Prices above average (BUY):", above_average)
print("Prices at or below average (WAIT):", at_or_below_average)

assert len(prices) == 8
assert round(average, 3) == 108.375
assert above_average == [110, 115, 120]
assert at_or_below_average == [100, 102, 105, 108, 107]

print("All checks passed!")
```
